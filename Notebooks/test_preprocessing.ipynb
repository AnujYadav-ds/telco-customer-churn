```python
from src.preprocessing import (
    load_data,
    clean_data,
    engineer_features,
    prepare_features_target,
    split_data,
    identify_feature_types,
    build_preprocessor
)
```

```python
df_test = load_data("../Data/WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Raw data shape:", df_test.shape)
```

```python
df_clean_test = clean_data(df_test)

print("Cleaned data shape:", df_clean_test.shape)
print("\nChurn distribution:")
print(df_clean_test["Churn"].value_counts())
print("\nMissing TotalCharges:", df_clean_test["TotalCharges"].isna().sum())
```

```python
df_engineered_test = engineer_features(df_clean_test)

print("Engineered data shape:", df_engineered_test.shape)

print("\nEngineered features:")
print(
    df_engineered_test[
        ["tenure", "tenure_group", "num_services"]
    ].head(10)
)

print("\nNum services distribution:")
print(
    df_engineered_test["num_services"]
    .value_counts()
    .sort_index()
)
```

```python
X_test, y_test = prepare_features_target(df_engineered_test)

print("X shape:", X_test.shape)
print("y shape:", y_test.shape)

print("\nTarget distribution:")
print(y_test.value_counts())
```

```python
X_train, X_test, y_train, y_test = split_data(X_test, y_test)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTest class distribution:")
print(y_test.value_counts())
```

```python
numeric_features, categorical_features = identify_feature_types(X_train)

print("Numeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
preprocessor_test = build_preprocessor(
    numeric_features,
    categorical_features
)

print(preprocessor_test)
```

```python
preprocessor_test.fit(X_train)

X_train_processed_test = preprocessor_test.transform(X_train)
X_test_processed_test = preprocessor_test.transform(X_test)

print("Processed X_train shape:", X_train_processed_test.shape)
print("Processed X_test shape:", X_test_processed_test.shape)
```
